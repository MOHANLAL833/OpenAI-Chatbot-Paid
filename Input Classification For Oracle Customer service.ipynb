```python
from  openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()
client=OpenAI(api_key=os.getenv("OPENAI_APIKEY"))
```

```python
def get_completion_message(messages,model='gpt-4o-mini',temperature=0,max_tokens=500):
    
       response=client.chat.completions.create(model=model,
                                               messages=messages,
                                               temperature=temperature,
                                               max_tokens=max_tokens)
       return response.choices[0].message.content
```

Classify customer queries to handle different Queries

```python
delimiter = "####"

system_message = f"""
You will be provided with Oracle Fusion SCM customer support queries.
Each query will be delimited with {delimiter} characters.

Your task is to classify each query into a primary category
and a secondary category.

Return your answer ONLY in JSON format with the following keys:
- primary
- secondary

Primary Categories:

1. Procurement
2. Order Management
3. Inventory Management
4. Manufacturing
5. Shipping & Logistics
6. Product Management
7. General Inquiry

Procurement Secondary Categories:
- Purchase Requisition
- Purchase Order Creation
- Supplier Management
- Approval Workflow
- Receiving
- Invoice Matching

Order Management Secondary Categories:
- Sales Order Creation
- Order Import
- Order Fulfillment
- Order Cancellation
- Pricing Issues
- Order Holds

Inventory Management Secondary Categories:
- Item Creation
- Stock Availability
- Inventory Transactions
- Cycle Count
- Lot and Serial Control
- Inventory Organization

Manufacturing Secondary Categories:
- Work Order Creation
- Production Scheduling
- Resource Assignment
- Material Issues
- Work Order Completion
- Manufacturing Execution

Shipping & Logistics Secondary Categories:
- Pick Release
- Shipping Confirmation
- Delivery Assignment
- Shipment Tracking
- Freight Charges
- Shipping Exceptions

Product Management Secondary Categories:
- Item Master
- Item Attributes
- Catalog Management
- BOM Management
- Product Lifecycle

General Inquiry Secondary Categories:
- Navigation Help
- Report Request
- User Access
- Training
- Speak to Functional Consultant
"""
```

```python
user_message=f"""\ I want to create a purchase order for the item"""
messages=[{'role':'system','content':system_message},
         {'role':'user','content':f"{delimiter}{user_message}{delimiter}"}]
response=get_completion_message(messages)
print(response)
```
