```python
from  openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()
client=OpenAI(api_key=os.getenv("OPENAI_APIKEY"))
```

Helper Function with Temperature setting

```python
def get_completion(prompt,model='gpt-4o-mini'):
    message=[{"role":"system","content":prompt}]
    #message.append({"role":"user","content":prompt})
    response=client.chat.completions.create(model=model,messages=message,temperature=0)
    # chat_gpt_reply=response.choices[0].message.content
    # message.append({"role":"assistant","content":chat_gpt_reply})
    return response.choices[0].message.content
```

```python
response=get_completion("What is the captial of Australia?")
```

```python
def get_completion_message(messages,model='gpt-4o-mini',temperature=0,max_tokens=500):
    
       response=client.chat.completions.create(model=model,
                                               messages=messages,
                                               temperature=temperature,
                                               max_tokens=max_tokens)
       return response.choices[0].message.content
```

```python
message=[{"role":"system",
         "content":"You are a great English writer"},
        {"role":"user",
         "content":"write a poem about a happy man"}]
response2=get_completion_message(message,temperature=1) # the temperature is set to 1 because we need to have creative answer
print(response2)
```

```python
message=[{"role":"system",
         "content":"All your response must be one scentence long"},
        {"role":"user",
         "content":"write a poem about a happy man"}]
response3=get_completion_message(message,temperature=1) # the temperature is set to 1 because we need to have creative answer
print(response3)
```

Token count

```python
def get_completion_message_token(messages,model='gpt-4o-mini',temperature=0,max_tokens=500):
    
       response=client.chat.completions.create(model=model,
                                               messages=messages,
                                               temperature=temperature,
                                               max_tokens=max_tokens)
       content=response.choices[0].message.content
       token_count={'input_token':response.usage.prompt_tokens,
                   'completion_token':response.usage.completion_tokens,
                   'total_tokens':response.usage.total_tokens}
       return content,token_count
```

```python
message=[{"role":"system",
         "content":"All your response must be one scentence long"},
        {"role":"user",
         "content":"write a poem about a happy man"}]
token_response=get_completion_message_token(message,temperature=1) # the temperature is set to 1 because we need to have creative answer
print(token_response)
```
