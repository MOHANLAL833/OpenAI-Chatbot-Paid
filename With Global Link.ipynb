```python
from  openai import OpenAI
from dotenv import load_dotenv
import gradio
import os

load_dotenv()
client=OpenAI(api_key=os.getenv("OPENAI_APIKEY"))
```

```python
message=[{"role":"system","content":"You are a skilled Pyhton developer"}]
```

```python
def custom_gpt(user_input):
    message.append({"role":"user","content":user_input})
    response=client.chat.completions.create(model='gpt-4o-mini',messages=message)
    chat_gpt_reply=response.choices[0].message.content
    message.append({"role":"assistant","content":chat_gpt_reply})
    return chat_gpt_reply
```

```python
demo=gradio.Interface(fn=custom_gpt,inputs="text",outputs="text",title="Pyhton development")
demo.launch(share=True)
```
